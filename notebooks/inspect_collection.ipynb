# Inspect SAE Feature Stability Collection

This notebook is a lightweight viewer for the descriptive inspection artifacts generated by `uv run run-inspection`. It does not run statistical tests or interpret SAE features.

```python
from pathlib import Path

import pandas as pd
from IPython.display import IFrame, Markdown, display

repo_root = Path.cwd().resolve()
if repo_root.name == "notebooks":
    repo_root = repo_root.parent

output_dir = repo_root / "output"
figures_dir = output_dir / "figures"
parquet_path = output_dir / "activations.parquet"
report_path = output_dir / "inspection_report.md"

assert parquet_path.exists()
assert report_path.exists()
assert figures_dir.exists()

df = pd.read_parquet(parquet_path, engine="pyarrow")
df.shape
```

## Dataset Snapshot

```python
display(df.head())
display(df.groupby("layer")[["activation_norm", "nonzero_features", "prompt_token_len"]].agg(["min", "mean", "median", "max"]))
```

## Figures

```python
for name in [
    "activation_norms_by_factor.pdf",
    "nonzero_features_by_factor.pdf",
    "feature_overlap_heatmap.pdf",
    "top_features_per_cell.pdf",
]:
    path = figures_dir / name
    assert path.exists(), path
    display(Markdown(f"### {name}"))
    display(IFrame(src=str(path.relative_to(repo_root)), width="100%", height=650))
```
